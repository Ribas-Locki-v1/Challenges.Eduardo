```python
def reverseCases(s: str) -> str:
    s = s.swapcase()
    return s
```

```python
print(reverseCases("may the force be with you"))
print(reverseCases("BOND, JAMES BOND"))
print(reverseCases("hOUSTON, wE hAVE A pROBLEM"))
print(reverseCases("I'LL BE back!"))
```

```python
def reverseCases(s: str) -> str:
    s2 = ""
    for c in s:
        if c.isupper():
            s2+=c.lower()
        else:
            s2+=c.upper()
    return s2
```

```python
print(reverseCases("may the force be with you"))
print(reverseCases("BOND, JAMES BOND"))
print(reverseCases("hOUSTON, wE hAVE A pROBLEM"))
print(reverseCases("I'LL BE back!"))
```
